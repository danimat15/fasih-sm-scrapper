```python
import pandas as pd
```

```python
df = pd.read_csv("update_data.csv",sep=",")
df.head(10)    
```

```python
df['Skala Usaha / Jenis Prelist'].unique()
```

```python
# Filter data berdasarkan 'Nama Keluarga/Bangunan/Usaha' yang mengandung kata 'kosong'
df['Nama Keluarga/Bangunan/Usaha'] = df['Nama Keluarga/Bangunan/Usaha'].fillna('').astype(str)

# Filter case-insensitive (mengabaikan huruf besar/kecil)
is_kosong = df['Nama Keluarga/Bangunan/Usaha'].str.contains('kosong', case=False)

df_kosong = df[is_kosong]
df_tidak_kosong = df[~is_kosong]

# Tampilkan ringkasan jumlah
print("==================================================")
print("ANALISIS BANGUNAN KOSONG")
print("==================================================")
print(f"Total Baris Data             : {len(df)}")
print(f"Jumlah Bangunan Kosong       : {len(df_kosong)}")
print(f"Jumlah Bangunan Tidak Kosong : {len(df_tidak_kosong)}")
print("==================================================")

# Tampilkan contoh data menggunakan display agar format tabel rapi di notebook
print("\n[CONTOH BANGUNAN KOSONG]")
display(df_kosong[['Kode Identitas', 'Nama Keluarga/Bangunan/Usaha']].head(10))

print("\n[CONTOH BANGUNAN TIDAK KOSONG]")
display(df_tidak_kosong[['Kode Identitas', 'Nama Keluarga/Bangunan/Usaha']].head(10))
```

```python
kode = df['Kode Identitas'].astype(str).str.ljust(16, 'X')

df['kab']     = kode.str[:4]
df['kec']     = kode.str[:7]
df['desa']    = kode.str[:10]
df['sls']     = kode.str[:14]
df['sub_sls'] = kode.str[:16]

df[['Kode Identitas', 'kab', 'kec', 'desa', 'sls', 'sub_sls']].head()
```

```python
levels = ['kab', 'kec', 'desa', 'sls', 'sub_sls']

tabulasi = {}
for lvl in levels:
    tab = (
        df.groupby([lvl, 'Skala Usaha / Jenis Prelist'])
        .size()
        .unstack(fill_value=0)
    )
    tab['Total'] = tab.sum(axis=1)
    tabulasi[lvl] = tab

# Tampilkan semua tabulasi
for lvl, tab in tabulasi.items():
    print(f"\n{'='*60}")
    print(f"Tabulasi berdasarkan: {lvl.upper()}")
    print('='*60)
    display(tab)
```
